```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1 import make_axes_locatable

from acgan import ACGAN
```

Loading model.

```python
acgan2 = ACGAN(device='cpu', latent_dim=2, lin_num_neurons_discriminator=[32])
acgan4 = ACGAN(device='cpu', latent_dim=4, lin_num_neurons_discriminator=[32])
acgan8 = ACGAN(device='cpu', latent_dim=8, lin_num_neurons_discriminator=[32])
acgan16 = ACGAN(device='cpu', latent_dim=16, lin_num_neurons_discriminator=[32])
acgan32 = ACGAN(device='cpu', latent_dim=32, lin_num_neurons_discriminator=[32])
acgan64 = ACGAN(device='cpu', latent_dim=64, lin_num_neurons_discriminator=[32])
acgan128 = ACGAN(device='cpu', latent_dim=128, lin_num_neurons_discriminator=[32])
```

```python
model_g2 = torch.load('best_models/acgan_ld2_ligcl_ausio2_epoch400.pth', map_location=torch.device('cpu'), weights_only=True)
model_d2 = torch.load('best_models/acgan_ld2_ligcl_ausio2_epoch400.pth', map_location=torch.device('cpu'), weights_only=True)
model_g4 = torch.load('best_models/acgan_ld4_ligcl_ausio2_epoch400.pth', map_location=torch.device('cpu'), weights_only=True)
model_d4 = torch.load('best_models/acgan_ld4_ligcl_ausio2_epoch400.pth', map_location=torch.device('cpu'), weights_only=True)
model_g8 = torch.load('best_models/acgan_ld8_ligcl_ausio2_epoch400.pth', map_location=torch.device('cpu'), weights_only=True)
model_d8 = torch.load('best_models/acgan_ld8_ligcl_ausio2_epoch400.pth', map_location=torch.device('cpu'), weights_only=True)
model_g16 = torch.load('best_models/acgan_ld16_ligcl_ausio2_epoch400.pth', map_location=torch.device('cpu'), weights_only=True)
model_d16 = torch.load('best_models/acgan_ld16_ligcl_ausio2_epoch400.pth', map_location=torch.device('cpu'), weights_only=True)
model_g32 = torch.load('best_models/acgan_ld32_ligcl_ausio2_epoch400.pth', map_location=torch.device('cpu'), weights_only=True)
model_d32 = torch.load('best_models/acgan_ld32_ligcl_ausio2_epoch400.pth', map_location=torch.device('cpu'), weights_only=True)
model_g64 = torch.load('best_models/acgan_ld64_ligcl_ausio2_epoch400.pth', map_location=torch.device('cpu'), weights_only=True)
model_d64 = torch.load('best_models/acgan_ld64_ligcl_ausio2_epoch400.pth', map_location=torch.device('cpu'), weights_only=True)
model_g128 = torch.load('best_models/acgan_ld128_ligcl_ausio2_epoch400.pth', map_location=torch.device('cpu'), weights_only=True)
model_d128 = torch.load('best_models/acgan_ld128_ligcl_ausio2_epoch400.pth', map_location=torch.device('cpu'), weights_only=True)

acgan2.generator.load_state_dict(model_g2['generator'])
acgan2.discriminator.load_state_dict(model_d2['discriminator'])
acgan4.generator.load_state_dict(model_g4['generator'])
acgan4.discriminator.load_state_dict(model_d4['discriminator'])
acgan8.generator.load_state_dict(model_g8['generator'])
acgan8.discriminator.load_state_dict(model_d8['discriminator'])
acgan16.generator.load_state_dict(model_g16['generator'])
acgan16.discriminator.load_state_dict(model_d16['discriminator'])
acgan32.generator.load_state_dict(model_g32['generator'])
acgan32.discriminator.load_state_dict(model_d32['discriminator'])
acgan64.generator.load_state_dict(model_g64['generator'])
acgan64.discriminator.load_state_dict(model_d64['discriminator'])
acgan128.generator.load_state_dict(model_g128['generator'])
acgan128.discriminator.load_state_dict(model_d128['discriminator'])
```

## Model avaliation

Loading data.

```python
data_ausio2 = np.load('../../autoencoders/dataset_TipAu_SiO2.npz')['vDeflection_N']
data_ligcl = np.load('../../autoencoders/dataset_TipLig_CL.npz')['vDeflection_N']

data_ausio2 = torch.tensor(data_ausio2, dtype=torch.float32).unsqueeze(1)
data_ausio2 /= data_ausio2.max()
data_ligcl = torch.tensor(data_ligcl, dtype=torch.float32).unsqueeze(1)
data_ligcl /= data_ligcl.max()
```

Testing.

```python
label_ligcl, label_ausio2 = torch.tensor([[1.0, 0.0]]), torch.tensor([[0.0, 1.0]])

sample2_ausio2 = torch.randn(1, 2)
sample2_ligcl = torch.randn(1, 2)
sample4_ausio2 = torch.randn(1, 4)
sample4_ligcl = torch.randn(1, 4)
sample8_ausio2 = torch.randn(1, 8)
sample8_ligcl = torch.randn(1, 8)
sample16_ausio2 = torch.randn(1, 16)
sample16_ligcl = torch.randn(1, 16)
sample32_ausio2 = torch.randn(1, 32)
sample32_ligcl = torch.randn(1, 32)
sample64_ausio2 = torch.randn(1, 64)
sample64_ligcl = torch.randn(1, 64)
sample128_ausio2 = torch.randn(1, 128)
sample128_ligcl = torch.randn(1, 128)

with torch.no_grad():
    gen2_ausio2 = acgan2.generator(sample2_ausio2, label_ausio2)
    gen2_ligcl = acgan2.generator(sample2_ligcl, label_ligcl)
    gen4_ausio2 = acgan4.generator(sample4_ausio2, label_ausio2)
    gen4_ligcl = acgan4.generator(sample4_ligcl, label_ligcl)
    gen8_ausio2 = acgan8.generator(sample8_ausio2, label_ausio2)
    gen8_ligcl = acgan8.generator(sample8_ligcl, label_ligcl)
    gen16_ausio2 = acgan16.generator(sample16_ausio2, label_ausio2)
    gen16_ligcl = acgan16.generator(sample16_ligcl, label_ligcl)
    gen32_ausio2 = acgan32.generator(sample32_ausio2, label_ausio2)
    gen32_ligcl = acgan32.generator(sample32_ligcl, label_ligcl)
    gen64_ausio2 = acgan64.generator(sample64_ausio2, label_ausio2)
    gen64_ligcl = acgan64.generator(sample64_ligcl, label_ligcl)
    gen128_ausio2 = acgan128.generator(sample128_ausio2, label_ausio2)
    gen128_ligcl = acgan128.generator(sample128_ligcl, label_ligcl)
    
fig, ax = plt.subplots(4, 2, figsize=(15, 10)) 

ax[0, 0].plot(data_ausio2[0].squeeze())
ax[0, 0].set_title('AuSiO2 Real')

ax[1, 0].plot(gen2_ausio2[0].squeeze())
ax[1, 0].set_title('AuSiO2 LD=2')

ax[2, 0].plot(gen4_ausio2[0].squeeze())
ax[2, 0].set_title('AuSiO2 LD=4')

ax[3, 0].plot(gen8_ausio2[0].squeeze())
ax[3, 0].set_title('AuSiO2 LD=8')

# LigCL Plots (Right Column)
ax[0, 1].plot(data_ligcl[2].squeeze())
ax[0, 1].set_title('LigCL Real')

ax[1, 1].plot(gen2_ligcl[0].squeeze())
ax[1, 1].set_title('LigCL LD=2')

ax[2, 1].plot(gen4_ligcl[0].squeeze())
ax[2, 1].set_title('LigCL LD=4')

ax[3, 1].plot(gen8_ligcl[0].squeeze())
ax[3, 1].set_title('LigCL LD=8')

plt.tight_layout()
plt.show()

fig, ax = plt.subplots(4, 2, figsize=(15, 10))

# AuSiO2 Plots (Left Column)
ax[0, 0].plot(gen16_ausio2[0].squeeze())
ax[0, 0].set_title('AuSiO2 LD=16')

ax[1, 0].plot(gen32_ausio2[0].squeeze())
ax[1, 0].set_title('AuSiO2 LD=32')

ax[2, 0].plot(gen64_ausio2[0].squeeze())
ax[2, 0].set_title('AuSiO2 LD=64')

ax[3, 0].plot(gen128_ausio2[0].squeeze())
ax[3, 0].set_title('AuSiO2 LD=128')

# LigCL Plots (Right Column)
ax[0, 1].plot(gen16_ligcl[0].squeeze())
ax[0, 1].set_title('LigCL LD=16')

ax[1, 1].plot(gen32_ligcl[0].squeeze())
ax[1, 1].set_title('LigCL LD=32')

ax[2, 1].plot(gen64_ligcl[0].squeeze())
ax[2, 1].set_title('LigCL LD=64')

ax[3, 1].plot(gen128_ligcl[0].squeeze())
ax[3, 1].set_title('LigCL LD=128')

plt.tight_layout()
plt.show()
```

## Classification

```python
_, cla1 = acgan2.discriminator(data_ausio2[800:])
_, cla0 = acgan2.discriminator(data_ligcl[800:])

pred1 = cla1.argmax(dim=1)  
pred0 = cla0.argmax(dim=1)  
true1 = torch.ones_like(pred1)
true0 = torch.zeros_like(pred0)
acc1 = (pred1 == true1).float().mean().item() * 100  
acc0 = (pred0 == true0).float().mean().item() * 100  

total_correct = (pred1 == true1).sum() + (pred0 == true0).sum()
total_samples = len(pred1) + len(pred0)
total_acc = (total_correct / total_samples).item() * 100

print('LD=2')
print(f"Class 1 Accuracy: {acc1:.2f}%")
print(f"Class 0 Accuracy: {acc0:.2f}%") 
print(f"Total Accuracy: {total_acc:.2f}%")
```
